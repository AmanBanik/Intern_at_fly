This lane answers a high-value, proactive business question. ## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**What decision does this improve?**
It improves the decision of *which* high-value pages the content team should review and prioritize for refresh or protection next.

**Who acts on the output, and what do they do?**
An editor or content manager will act on the ranked list by reviewing the flagged pages and making targeted updates (e.g., expanding content, updating metadata) to protect against the predicted decline.

**What does a wrong answer cost?**
A false positive costs the editor's time, as they might review a page that was naturally going to recover. A false negative is more costly—it means missing a high-value page that then suffers a severe traffic drop, costing the client revenue and visibility.

**Why does data or ML help?**
A simple rule (e.g., 'it dropped yesterday') is purely reactive. Machine learning can weigh multiple interconnected signals (impressions, CTR gaps, position decay, and query concentration) over prior time windows to identify leading indicators of a decline before it fully materializes.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Calculate a few quick stats
total_pages = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
high_volume_declining = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] > 1000)])

print(f"Total content items in starter dataset: {total_pages:,}")
print(f"Pages currently showing a 'down' trend: {declining_pages:,} ({declining_pages/total_pages:.1%})")
print(f"High-volume declining pages (>1000 impressions in 90d): {high_volume_declining:,}")
print(f"\nTakeaway: Over {declining_pages/total_pages:.0%} of the content is already showing decline signals, and {high_volume_declining:,} of those are high-volume pages. Building a predictive model to flag these *before* their downward trend becomes severe represents a massive opportunity to save traffic.")
```
