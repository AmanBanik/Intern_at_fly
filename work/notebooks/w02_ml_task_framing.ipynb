## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a **Classification** task that outputs probabilities, ultimately used for **Ranking / Scoring**.

While the strict target is binary (Will it decline: Yes/No), we care less about a hard Yes/No and more about the *probability* of decline (e.g., 85% risk). Sorting the pages by this predicted probability allows us to create a prioritized, ranked review queue for the content team.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Target:** A binary label representing an **observed future outcome** (e.g., `future_decline_label`).

**Where it comes from:** It is derived strictly from real, observed search performance data (impressions/clicks), not from a predefined product rule or health score. Specifically, we will calculate if a page's impressions in a *future window* (e.g., next 30 days) drop by more than a set threshold (e.g., >15% relative drop AND >100 absolute drop) compared to a *prior feature window* (e.g., last 90 days).

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Success Metric:** **Precision@K** (e.g., Precision@50) and **Average Precision**.

**What 'good' means:** Overall accuracy is misleading for an imbalanced risk-prediction dataset. The content team has limited capacity to review pages. 'Good' means that out of the top `K` (e.g., 50) pages the model flags as having the highest risk, a very high percentage actually went on to decline (e.g., Precision@50 > 0.60). We want to maximize the efficiency of the human reviewer's time.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (e.g., `if traffic_dropped_last_week: review`) is **reactive**—it only catches the fire after the traffic has already burned down. 

Machine learning is **proactive**. It can weigh the tangled, non-linear interactions of multiple *leading indicators* (e.g., a page has high age + slowly decaying CTR + highly concentrated query mix). A simple if-statement cannot effectively combine 15 different weak signals to detect hidden momentum shifts, but an ML model like a Random Forest excels at finding these complex, multi-dimensional patterns *before* the steep drop occurs.
